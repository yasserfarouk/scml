```python
from scml.oneshot.context import (
    EutopiaConsumerContext,
    EutopiaSupplierContext,
    SingleAgentPerLevelConsumerContext,
    SingleAgentPerLevelSupplierContext,
)

from scml.oneshot.agents import EqualDistOneShotAgent

from scml.runner import WorldRunner

from scml.oneshot import builtin_agent_types as builtin_oneshot
from scml.std import builtin_agent_types as builtin_std
from scml.std import StdWorld
```

```python
runner = WorldRunner(SingleAgentPerLevelConsumerContext())

agents = builtin_oneshot()

for agent in agents:
    runner.add(agent)

runner.runall()
```

```python
runner.plot_stats(agg=False, ylegend=2, legend_ncols=4);
```

```python
runner = WorldRunner(SingleAgentPerLevelConsumerContext(world_type=StdWorld, perishable=False))

agents = builtin_std()

for agent in agents:
    runner.add(agent)

runner.runall()
```

```python
runner.plot_stats(agg=False, ylegend=2, legend_ncols=4);
```
